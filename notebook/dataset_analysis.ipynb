```python
def prepare_train_test_split(
    df,
    num_cv=10,
    block=9,
    min_samples=20,
    owner_col="owner",
    verbose=True,
):
    """
    Create block-based train/test split, keep only active developers in train,
    and remove unseen developers from test.
    """
    samples_per_block = len(df) // num_cv
    sliced_df = df[: samples_per_block * (block + 1)]

    df_train = sliced_df[: samples_per_block * block].copy()
    df_test = sliced_df[samples_per_block * block : samples_per_block * (block + 1)].copy()

    developers = df_train[owner_col].value_counts()
    filtered_developers = developers.index[developers >= min_samples]
    df_train = df_train[df_train[owner_col].isin(filtered_developers)]

    train_owners = set(df_train[owner_col])
    test_owners = set(df_test[owner_col])
    unseen_test_owners = list(test_owners - train_owners)

    df_test = df_test[~df_test[owner_col].isin(unseen_test_owners)]

    if verbose:
        print(f"Samples per block: {samples_per_block}, Selected block: {block}")
        print(f"Training data: {len(df_train)}, Validation data: {len(df_test)}")
        print(f"Number of developers in train: {len(df_train[owner_col].unique())}")
        print(f"Number of developers in test: {len(df_test[owner_col].unique())}")
        print(f"Train dataset size: {len(df_train)}")
        print(f"Test dataset size: {len(df_test)}")

    return df_train, df_test
```

```python
# Apply the reusable split function to all loaded dataframes
dataset_frames = {
    "typescript": ts_df,
    "openj9": oj9_df,
    "google_chromium": gc_df,
    "mozilla_firefox": mf_df,
    "mozilla_core": mc_df,
}

dataset_splits = {}
for name, df_ in dataset_frames.items():
    print(f"\n=== {name} ===")
    train_df, test_df = prepare_train_test_split(
        df_,
        num_cv=10,
        block=9,
        min_samples=20,
        owner_col="owner",
        verbose=True,
    )
    dataset_splits[name] = {"train": train_df, "test": test_df}

# Optional direct variables for convenience
ts_train, ts_test = dataset_splits["typescript"]["train"], dataset_splits["typescript"]["test"]
oj9_train, oj9_test = dataset_splits["openj9"]["train"], dataset_splits["openj9"]["test"]
gc_train, gc_test = dataset_splits["google_chromium"]["train"], dataset_splits["google_chromium"]["test"]
mf_train, mf_test = dataset_splits["mozilla_firefox"]["train"], dataset_splits["mozilla_firefox"]["test"]
mc_train, mc_test = dataset_splits["mozilla_core"]["train"], dataset_splits["mozilla_core"]["test"]
```
